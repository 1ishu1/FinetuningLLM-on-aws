```python

dataset_path = "s3://llm-finetune-dataset-ishu/datasets/pharma_instruction_data.csv/"
```

```python
from datasets import load_dataset
```

```python
dataset = load_dataset("csv",data_files={"train":dataset_path},split="train")
dataset
```

```python
dataset[0]
```

```python
def format_example(example):
    prompt = f"### Instruction:\n{example['instruction']}\n### Input:\n{example['input']}\n### Response:\n{example['output']}"
    return {"text": prompt}
```

```python
dataset = dataset.map(format_example)
```

```python
dataset["text"][0]
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, TrainingArguments, Trainer
```

```python
model_id = "TinyLlama/TinyLlama-1.1B-intermediate-step-1431k-3T"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
```

```python
def tokenize_fn(example):
    tokens = tokenizer(example["text"], truncation=True, padding="max_length", max_length=512)
    tokens["labels"] = tokens["input_ids"].copy()
    return tokens
```

```python
tokenized = dataset.map(tokenize_fn, batched=True)
```

```python
tokenized
```

```python
tokenized["input_ids"] # 2 are pads/eos(end of sentence)
```

```python
tokenized["attention_mask"] # 1 means useful
```

```python
tokenized["labels"]   # labels and input_ids are same as since it is auto-regressive
                        #(convert instr./input/output in such dataset where combined instr.+some part of i/p will predict next part of i/p) model
```

```python
from peft import LoraConfig, get_peft_model, TaskType
```

```python
model = AutoModelForCausalLM.from_pretrained(model_id, device_map="auto")  #aws not allow to load model inside the jupyter
```

```python
lora_config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    r=8,
    lora_alpha=16,
    lora_dropout=0.05,
    target_modules=["q_proj", "v_proj"],
    bias="none"
)
```

```python
model = get_peft_model(model, lora_config)
```

```python
args = TrainingArguments(
    output_dir="./tinyllama-instruction",
    num_train_epochs=3,
    per_device_train_batch_size=1,
    gradient_accumulation_steps=8,
    learning_rate=2e-4,
    fp16=True,
    logging_steps=20,
    save_total_limit=1,
    report_to="none"
)
```

```python
trainer = Trainer(
    model=model,
    args=args,
    train_dataset=tokenized,
)
```

```python
trainer.train()
```

```python
model_path = "/content/tinyllama-instruction/checkpoint-3"
```

```python
instruction_model = AutoModelForCausalLM.from_pretrained(model_path, device_map="auto")
```

```python
prompt = "Explain the mechanism of action of Metformin."
```

```python
inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
```

```python
outputs = instruction_model.generate(
    **inputs,
    max_new_tokens=100,
    temperature=0.8,
    top_p=0.9,
    do_sample=True,
    repetition_penalty=1.1
)
```

```python
print("\nModel Output:\n")
```

```python
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
